## Phase 8B.2 — Three-step reversal and parity validation

The three-step reversal and parity-coordinate definitions were implemented and
protected by synthetic unit tests before being applied to the real trajectory
data.

For

\[
S_3=(\mathbf d_1,\mathbf d_2,\mathbf d_3),
\]

the exact reversed sequence is

\[
R(S_3)=(-\mathbf d_3,-\mathbf d_2,-\mathbf d_1).
\]

The pre-specified parity coordinates are

\[
\mathbf n=\mathbf d_1+\mathbf d_2+\mathbf d_3,
\]

\[
\mathbf q=\mathbf d_1-2\mathbf d_2+\mathbf d_3,
\]

and

\[
\mathbf e=\mathbf d_1-\mathbf d_3.
\]

Under exact reversal,

\[
\mathbf n\rightarrow-\mathbf n,
\qquad
\mathbf q\rightarrow-\mathbf q,
\qquad
\mathbf e\rightarrow\mathbf e.
\]

The transformation is invertible, so the full six-dimensional three-step
sequence can be reconstructed exactly from \((\mathbf n,\mathbf q,\mathbf e)\).

This section performs only structural and algebraic validation on the real
data. No classifier performance or inferential statistic is evaluated yet.

## Reconstructing the co-moving representation

TrackMate already has a historically validated saved co-moving trajectory
table.

CTC reference tracking does not require a separate committed co-moving data
file.

Instead, Phase 8B will reconstruct co-moving coordinates using the same
frame-wise common-motion definition used previously.

Before applying that reconstruction to CTC, the transformation will be
regression-checked against the saved TrackMate co-moving trajectory table.

The required validation is:

> applying the reconstructed common-motion transformation to raw TrackMate
> trajectories must reproduce the historically saved TrackMate co-moving
> coordinates.

Only after this check passes will the same transformation be applied to the
CTC reference trajectories.

This prevents an accidental change in the definition of the co-moving
representation during Phase 8B.

```python
from src.steps import compute_steps_for_tau


def compute_framewise_common_motion(
    tracks,
):
    """
    Estimate frame-wise population common motion
    from exact one-frame displacements.

    For each frame transition, common dx and dy
    are the component-wise medians across cells.
    """

    one_frame_steps = (
        compute_steps_for_tau(
            tracks,
            tau_frames=1,
        )
    )

    common_motion = (
        one_frame_steps
        .groupby(
            [
                "frame_start",
                "frame_end",
            ],
            as_index=False,
        )
        .agg(
            common_dx_um=(
                "dx_um",
                "median",
            ),
            common_dy_um=(
                "dy_um",
                "median",
            ),
            n_cells=(
                "cell_id",
                "nunique",
            ),
        )
        .sort_values(
            "frame_start"
        )
        .reset_index(
            drop=True
        )
    )

    return common_motion
```

```python
def make_comoving_tracks(
    tracks,
    common_motion,
):
    """
    Subtract cumulative frame-wise population
    common translation from trajectory coordinates.

    Frame 0 has zero cumulative correction.
    """

    frame_correction = pd.DataFrame(
        {
            "frame": [0],
            "common_x_um": [0.0],
            "common_y_um": [0.0],
        }
    )

    transition_correction = pd.DataFrame(
        {
            "frame":
                common_motion[
                    "frame_end"
                ].to_numpy(),

            "common_x_um":
                np.cumsum(
                    common_motion[
                        "common_dx_um"
                    ].to_numpy()
                ),

            "common_y_um":
                np.cumsum(
                    common_motion[
                        "common_dy_um"
                    ].to_numpy()
                ),
        }
    )

    frame_correction = pd.concat(
        [
            frame_correction,
            transition_correction,
        ],
        ignore_index=True,
    )

    result = (
        tracks
        .merge(
            frame_correction,
            on="frame",
            how="left",
            validate="many_to_one",
        )
        .copy()
    )

    if (
        result[
            [
                "common_x_um",
                "common_y_um",
            ]
        ]
        .isna()
        .any()
        .any()
    ):
        raise RuntimeError(
            "Missing cumulative common-motion "
            "correction for one or more frames."
        )

    result[
        "x_um"
    ] = (
        result["x_um"]
        - result["common_x_um"]
    )

    result[
        "y_um"
    ] = (
        result["y_um"]
        - result["common_y_um"]
    )

    return (
        result[
            [
                "cell_id",
                "frame",
                "t_min",
                "x_um",
                "y_um",
            ]
        ]
        .sort_values(
            [
                "cell_id",
                "frame",
            ]
        )
        .reset_index(
            drop=True
        )
    )
```

```python
trackmate_common_motion_phase8b = (
    compute_framewise_common_motion(
        trackmate_raw
    )
)


trackmate_comoving_reconstructed = (
    make_comoving_tracks(
        trackmate_raw,
        trackmate_common_motion_phase8b,
    )
)


trackmate_comoving_expected = (
    trackmate_comoving[
        [
            "cell_id",
            "frame",
            "t_min",
            "x_um",
            "y_um",
        ]
    ]
    .sort_values(
        [
            "cell_id",
            "frame",
        ]
    )
    .reset_index(
        drop=True
    )
)


print(
    "TrackMate common-motion transitions:",
    len(
        trackmate_common_motion_phase8b
    ),
)


print(
    "Maximum x reconstruction difference:",
    np.max(
        np.abs(
            trackmate_comoving_reconstructed[
                "x_um"
            ].to_numpy()
            - trackmate_comoving_expected[
                "x_um"
            ].to_numpy()
        )
    ),
)

print(
    "Maximum y reconstruction difference:",
    np.max(
        np.abs(
            trackmate_comoving_reconstructed[
                "y_um"
            ].to_numpy()
            - trackmate_comoving_expected[
                "y_um"
            ].to_numpy()
        )
    ),
)


assert np.array_equal(
    trackmate_comoving_reconstructed[
        [
            "cell_id",
            "frame",
        ]
    ].to_numpy(),
    trackmate_comoving_expected[
        [
            "cell_id",
            "frame",
        ]
    ].to_numpy(),
)


assert np.allclose(
    trackmate_comoving_reconstructed[
        [
            "x_um",
            "y_um",
        ]
    ].to_numpy(),
    trackmate_comoving_expected[
        [
            "x_um",
            "y_um",
        ]
    ].to_numpy(),
    atol=1e-10,
)


print(
    "TrackMate co-moving reconstruction regression: PASSED"
)
```

## Phase 8B.3 — Pre-classification feature-space and fold audit

The real-data three-step reversal and parity algebra has now been validated in
both tracking sources.

Before inspecting any classifier performance, the exact feature
representations and cross-validation folds are fixed.

The five pre-specified three-step feature spaces are:

### 1. `net_only`

$$
\mathbf{n}
=
\mathbf{d}_1+\mathbf{d}_2+\mathbf{d}_3
$$

This is the net displacement across the complete three-step interval.

---

### 2. `internal_odd_only`

$$
\mathbf{q}
=
\mathbf{d}_1
-
2\mathbf{d}_2
+
\mathbf{d}_3
$$

This reversal-odd coordinate depends on the internal distribution of motion
across the three ordered steps.

---

### 3. `even_only`

$$
\mathbf{e}
=
\mathbf{d}_1-\mathbf{d}_3
$$

This coordinate is invariant under exact time reversal and therefore serves as
a negative-control feature space.

---

### 4. `odd_combined`

$$
[\mathbf{n},\mathbf{q}]
$$

This four-dimensional representation contains both the net directional
component and the internal reversal-odd component.

---

### 5. `full_6d`

$$
[
\mathbf{d}_1,
\mathbf{d}_2,
\mathbf{d}_3
]
$$

This is the original physical six-dimensional displacement representation:

`(dx1, dy1, dx2, dy2, dx3, dy3)`

The full feature space will use the original displacement coordinates rather
than the equivalent parity-coordinate representation.

Although the two full representations are related by an invertible linear
transformation and therefore contain the same information, a regularized
classifier combined with feature scaling need not produce numerically
identical fitted models under an arbitrary change of coordinates.

Therefore, the original displacement coordinates are fixed here, before any
Phase 8B classifier AUC is inspected.

The cross-validation design remains the pre-specified `3-fold GroupKFold`,
with trajectory identity as the grouping variable.

No classifier is fitted in this section.

```python
def make_three_step_feature_spaces(
    forward_array,
):
    """
    Construct the five pre-specified Phase 8B
    three-step feature spaces and their exact
    reversed counterparts.
    """

    reverse_array = (
        reverse_k_step_sequences(
            forward_array
        )
    )

    (
        net_forward,
        internal_forward,
        even_forward,
    ) = decompose_three_step_parity(
        forward_array
    )

    (
        net_reverse,
        internal_reverse,
        even_reverse,
    ) = decompose_three_step_parity(
        reverse_array
    )

    forward_spaces = {
        "net_only":
            net_forward,

        "internal_odd_only":
            internal_forward,

        "even_only":
            even_forward,

        "odd_combined":
            np.column_stack(
                [
                    net_forward,
                    internal_forward,
                ]
            ),

        "full_6d":
            forward_array,
    }

    reverse_spaces = {
        "net_only":
            net_reverse,

        "internal_odd_only":
            internal_reverse,

        "even_only":
            even_reverse,

        "odd_combined":
            np.column_stack(
                [
                    net_reverse,
                    internal_reverse,
                ]
            ),

        "full_6d":
            reverse_array,
    }

    return (
        forward_spaces,
        reverse_spaces,
    )
```

```python
from sklearn.model_selection import (
    GroupKFold,
)


PHASE8B_N_SPLITS = 3


def audit_group_folds(
    X,
    y,
    groups,
    cv,
):
    """
    Audit grouped cross-validation structure
    without fitting a classifier.
    """

    records = []

    for (
        fold_id,
        (
            train_indices,
            test_indices,
        ),
    ) in enumerate(
        cv.split(
            X,
            y,
            groups,
        ),
        start=1,
    ):

        train_groups = np.unique(
            groups[
                train_indices
            ]
        )

        test_groups = np.unique(
            groups[
                test_indices
            ]
        )

        overlap = np.intersect1d(
            train_groups,
            test_groups,
        )

        records.append(
            {
                "fold":
                    fold_id,

                "n_train_examples":
                    len(
                        train_indices
                    ),

                "n_test_examples":
                    len(
                        test_indices
                    ),

                "n_train_groups":
                    len(
                        train_groups
                    ),

                "n_test_groups":
                    len(
                        test_groups
                    ),

                "train_label_fraction":
                    float(
                        y[
                            train_indices
                        ].mean()
                    ),

                "test_label_fraction":
                    float(
                        y[
                            test_indices
                        ].mean()
                    ),

                "group_overlap":
                    len(
                        overlap
                    ),
            }
        )

    return pd.DataFrame(
        records
    )
```

```python
ctc_phase8b_cv = GroupKFold(
    n_splits=PHASE8B_N_SPLITS
)

trackmate_phase8b_cv = GroupKFold(
    n_splits=PHASE8B_N_SPLITS
)


ctc_fold_audit = (
    audit_group_folds(
        X=ctc_raw_classifier_sets[
            "net_only"
        ]["X"],
        y=ctc_raw_classifier_sets[
            "net_only"
        ]["y"],
        groups=ctc_raw_classifier_sets[
            "net_only"
        ]["groups"],
        cv=ctc_phase8b_cv,
    )
)


trackmate_fold_audit = (
    audit_group_folds(
        X=trackmate_raw_classifier_sets[
            "net_only"
        ]["X"],
        y=trackmate_raw_classifier_sets[
            "net_only"
        ]["y"],
        groups=trackmate_raw_classifier_sets[
            "net_only"
        ]["groups"],
        cv=trackmate_phase8b_cv,
    )
)


print(
    "CTC 3-fold audit:"
)

display(
    ctc_fold_audit
)


print(
    "TrackMate 3-fold audit:"
)

display(
    trackmate_fold_audit
)
```
