## Phase 8B.2 — Three-step reversal and parity validation

The three-step reversal and parity-coordinate definitions were implemented and
protected by synthetic unit tests before being applied to the real trajectory
data.

For

\[
S_3=(\mathbf d_1,\mathbf d_2,\mathbf d_3),
\]

the exact reversed sequence is

\[
R(S_3)=(-\mathbf d_3,-\mathbf d_2,-\mathbf d_1).
\]

The pre-specified parity coordinates are

\[
\mathbf n=\mathbf d_1+\mathbf d_2+\mathbf d_3,
\]

\[
\mathbf q=\mathbf d_1-2\mathbf d_2+\mathbf d_3,
\]

and

\[
\mathbf e=\mathbf d_1-\mathbf d_3.
\]

Under exact reversal,

\[
\mathbf n\rightarrow-\mathbf n,
\qquad
\mathbf q\rightarrow-\mathbf q,
\qquad
\mathbf e\rightarrow\mathbf e.
\]

The transformation is invertible, so the full six-dimensional three-step
sequence can be reconstructed exactly from \((\mathbf n,\mathbf q,\mathbf e)\).

This section performs only structural and algebraic validation on the real
data. No classifier performance or inferential statistic is evaluated yet.

## Reconstructing the co-moving representation

TrackMate already has a historically validated saved co-moving trajectory
table.

CTC reference tracking does not require a separate committed co-moving data
file.

Instead, Phase 8B will reconstruct co-moving coordinates using the same
frame-wise common-motion definition used previously.

Before applying that reconstruction to CTC, the transformation will be
regression-checked against the saved TrackMate co-moving trajectory table.

The required validation is:

> applying the reconstructed common-motion transformation to raw TrackMate
> trajectories must reproduce the historically saved TrackMate co-moving
> coordinates.

Only after this check passes will the same transformation be applied to the
CTC reference trajectories.

This prevents an accidental change in the definition of the co-moving
representation during Phase 8B.

```python
from src.steps import compute_steps_for_tau


def compute_framewise_common_motion(
    tracks,
):
    """
    Estimate frame-wise population common motion
    from exact one-frame displacements.

    For each frame transition, common dx and dy
    are the component-wise medians across cells.
    """

    one_frame_steps = (
        compute_steps_for_tau(
            tracks,
            tau_frames=1,
        )
    )

    common_motion = (
        one_frame_steps
        .groupby(
            [
                "frame_start",
                "frame_end",
            ],
            as_index=False,
        )
        .agg(
            common_dx_um=(
                "dx_um",
                "median",
            ),
            common_dy_um=(
                "dy_um",
                "median",
            ),
            n_cells=(
                "cell_id",
                "nunique",
            ),
        )
        .sort_values(
            "frame_start"
        )
        .reset_index(
            drop=True
        )
    )

    return common_motion
```

```python
def make_comoving_tracks(
    tracks,
    common_motion,
):
    """
    Subtract cumulative frame-wise population
    common translation from trajectory coordinates.

    Frame 0 has zero cumulative correction.
    """

    frame_correction = pd.DataFrame(
        {
            "frame": [0],
            "common_x_um": [0.0],
            "common_y_um": [0.0],
        }
    )

    transition_correction = pd.DataFrame(
        {
            "frame":
                common_motion[
                    "frame_end"
                ].to_numpy(),

            "common_x_um":
                np.cumsum(
                    common_motion[
                        "common_dx_um"
                    ].to_numpy()
                ),

            "common_y_um":
                np.cumsum(
                    common_motion[
                        "common_dy_um"
                    ].to_numpy()
                ),
        }
    )

    frame_correction = pd.concat(
        [
            frame_correction,
            transition_correction,
        ],
        ignore_index=True,
    )

    result = (
        tracks
        .merge(
            frame_correction,
            on="frame",
            how="left",
            validate="many_to_one",
        )
        .copy()
    )

    if (
        result[
            [
                "common_x_um",
                "common_y_um",
            ]
        ]
        .isna()
        .any()
        .any()
    ):
        raise RuntimeError(
            "Missing cumulative common-motion "
            "correction for one or more frames."
        )

    result[
        "x_um"
    ] = (
        result["x_um"]
        - result["common_x_um"]
    )

    result[
        "y_um"
    ] = (
        result["y_um"]
        - result["common_y_um"]
    )

    return (
        result[
            [
                "cell_id",
                "frame",
                "t_min",
                "x_um",
                "y_um",
            ]
        ]
        .sort_values(
            [
                "cell_id",
                "frame",
            ]
        )
        .reset_index(
            drop=True
        )
    )
```

```python
trackmate_common_motion_phase8b = (
    compute_framewise_common_motion(
        trackmate_raw
    )
)


trackmate_comoving_reconstructed = (
    make_comoving_tracks(
        trackmate_raw,
        trackmate_common_motion_phase8b,
    )
)


trackmate_comoving_expected = (
    trackmate_comoving[
        [
            "cell_id",
            "frame",
            "t_min",
            "x_um",
            "y_um",
        ]
    ]
    .sort_values(
        [
            "cell_id",
            "frame",
        ]
    )
    .reset_index(
        drop=True
    )
)


print(
    "TrackMate common-motion transitions:",
    len(
        trackmate_common_motion_phase8b
    ),
)


print(
    "Maximum x reconstruction difference:",
    np.max(
        np.abs(
            trackmate_comoving_reconstructed[
                "x_um"
            ].to_numpy()
            - trackmate_comoving_expected[
                "x_um"
            ].to_numpy()
        )
    ),
)

print(
    "Maximum y reconstruction difference:",
    np.max(
        np.abs(
            trackmate_comoving_reconstructed[
                "y_um"
            ].to_numpy()
            - trackmate_comoving_expected[
                "y_um"
            ].to_numpy()
        )
    ),
)


assert np.array_equal(
    trackmate_comoving_reconstructed[
        [
            "cell_id",
            "frame",
        ]
    ].to_numpy(),
    trackmate_comoving_expected[
        [
            "cell_id",
            "frame",
        ]
    ].to_numpy(),
)


assert np.allclose(
    trackmate_comoving_reconstructed[
        [
            "x_um",
            "y_um",
        ]
    ].to_numpy(),
    trackmate_comoving_expected[
        [
            "x_um",
            "y_um",
        ]
    ].to_numpy(),
    atol=1e-10,
)


print(
    "TrackMate co-moving reconstruction regression: PASSED"
)
```

## Phase 8B.3 — Pre-classification feature-space and fold audit

The real-data three-step reversal and parity algebra has now been validated in
both tracking sources.

Before inspecting any classifier performance, the exact feature
representations and cross-validation folds are fixed.

The five pre-specified three-step feature spaces are:

### 1. `net_only`

$$
\mathbf{n}
=
\mathbf{d}_1+\mathbf{d}_2+\mathbf{d}_3
$$

This is the net displacement across the complete three-step interval.

---

### 2. `internal_odd_only`

$$
\mathbf{q}
=
\mathbf{d}_1
-
2\mathbf{d}_2
+
\mathbf{d}_3
$$

This reversal-odd coordinate depends on the internal distribution of motion
across the three ordered steps.

---

### 3. `even_only`

$$
\mathbf{e}
=
\mathbf{d}_1-\mathbf{d}_3
$$

This coordinate is invariant under exact time reversal and therefore serves as
a negative-control feature space.

---

### 4. `odd_combined`

$$
[\mathbf{n},\mathbf{q}]
$$

This four-dimensional representation contains both the net directional
component and the internal reversal-odd component.

---

### 5. `full_6d`

$$
[
\mathbf{d}_1,
\mathbf{d}_2,
\mathbf{d}_3
]
$$

This is the original physical six-dimensional displacement representation:

`(dx1, dy1, dx2, dy2, dx3, dy3)`

The full feature space will use the original displacement coordinates rather
than the equivalent parity-coordinate representation.

Although the two full representations are related by an invertible linear
transformation and therefore contain the same information, a regularized
classifier combined with feature scaling need not produce numerically
identical fitted models under an arbitrary change of coordinates.

Therefore, the original displacement coordinates are fixed here, before any
Phase 8B classifier AUC is inspected.

The cross-validation design remains the pre-specified `3-fold GroupKFold`,
with trajectory identity as the grouping variable.

No classifier is fitted in this section.

```python
def make_three_step_feature_spaces(
    forward_array,
):
    """
    Construct the five pre-specified Phase 8B
    three-step feature spaces and their exact
    reversed counterparts.
    """

    reverse_array = (
        reverse_k_step_sequences(
            forward_array
        )
    )

    (
        net_forward,
        internal_forward,
        even_forward,
    ) = decompose_three_step_parity(
        forward_array
    )

    (
        net_reverse,
        internal_reverse,
        even_reverse,
    ) = decompose_three_step_parity(
        reverse_array
    )

    forward_spaces = {
        "net_only":
            net_forward,

        "internal_odd_only":
            internal_forward,

        "even_only":
            even_forward,

        "odd_combined":
            np.column_stack(
                [
                    net_forward,
                    internal_forward,
                ]
            ),

        "full_6d":
            forward_array,
    }

    reverse_spaces = {
        "net_only":
            net_reverse,

        "internal_odd_only":
            internal_reverse,

        "even_only":
            even_reverse,

        "odd_combined":
            np.column_stack(
                [
                    net_reverse,
                    internal_reverse,
                ]
            ),

        "full_6d":
            reverse_array,
    }

    return (
        forward_spaces,
        reverse_spaces,
    )
```

```python
from sklearn.model_selection import (
    GroupKFold,
)


PHASE8B_N_SPLITS = 3


def audit_group_folds(
    X,
    y,
    groups,
    cv,
):
    """
    Audit grouped cross-validation structure
    without fitting a classifier.
    """

    records = []

    for (
        fold_id,
        (
            train_indices,
            test_indices,
        ),
    ) in enumerate(
        cv.split(
            X,
            y,
            groups,
        ),
        start=1,
    ):

        train_groups = np.unique(
            groups[
                train_indices
            ]
        )

        test_groups = np.unique(
            groups[
                test_indices
            ]
        )

        overlap = np.intersect1d(
            train_groups,
            test_groups,
        )

        records.append(
            {
                "fold":
                    fold_id,

                "n_train_examples":
                    len(
                        train_indices
                    ),

                "n_test_examples":
                    len(
                        test_indices
                    ),

                "n_train_groups":
                    len(
                        train_groups
                    ),

                "n_test_groups":
                    len(
                        test_groups
                    ),

                "train_label_fraction":
                    float(
                        y[
                            train_indices
                        ].mean()
                    ),

                "test_label_fraction":
                    float(
                        y[
                            test_indices
                        ].mean()
                    ),

                "group_overlap":
                    len(
                        overlap
                    ),
            }
        )

    return pd.DataFrame(
        records
    )
```

```python
ctc_phase8b_cv = GroupKFold(
    n_splits=PHASE8B_N_SPLITS
)

trackmate_phase8b_cv = GroupKFold(
    n_splits=PHASE8B_N_SPLITS
)


ctc_fold_audit = (
    audit_group_folds(
        X=ctc_raw_classifier_sets[
            "net_only"
        ]["X"],
        y=ctc_raw_classifier_sets[
            "net_only"
        ]["y"],
        groups=ctc_raw_classifier_sets[
            "net_only"
        ]["groups"],
        cv=ctc_phase8b_cv,
    )
)


trackmate_fold_audit = (
    audit_group_folds(
        X=trackmate_raw_classifier_sets[
            "net_only"
        ]["X"],
        y=trackmate_raw_classifier_sets[
            "net_only"
        ]["y"],
        groups=trackmate_raw_classifier_sets[
            "net_only"
        ]["groups"],
        cv=trackmate_phase8b_cv,
    )
)


print(
    "CTC 3-fold audit:"
)

display(
    ctc_fold_audit
)


print(
    "TrackMate 3-fold audit:"
)

display(
    trackmate_fold_audit
)
```

### Paired orientation-randomization null

The null preserves every observed forward/reversed sequence pair.

For each null replicate, every pair is independently assigned a probability of
0.5 of being orientation-swapped.

The same swap mask is used across all five feature spaces within a replicate.

This preserves pairing between model performances and permits direct null
calibration of contrasts such as:

$$
AUC_{\mathrm{odd\ combined}}
-
AUC_{\mathrm{net\ only}}.
$$

The number of replicates and random seed were fixed before outcome inspection:

- null replicates: 200;
- logistic null seed: 2029.

## Phase 8B.5 — TrackMate tracking-source sensitivity

The primary CTC co-moving analysis and the supporting CTC raw analysis are now
complete.

The same pre-specified three-step logistic analysis is next applied to
TrackMate trajectories as a tracking-source sensitivity analysis.

CTC and TrackMate originate from the same microscopy sequence and are not
independent biological replicates.

Agreement between them therefore evaluates robustness to trajectory
reconstruction rather than biological replication.

No feature definition, cross-validation rule, classifier setting, null model,
or success criterion is changed for the TrackMate analysis.

```python
def run_phase8b_logistic_feature_analysis(
    forward_spaces,
    reverse_spaces,
    sequence_groups,
    classifier_sets,
    cv,
    null_seed,
    n_null=200,
):
    """
    Run the already pre-specified Phase 8B
    logistic analysis for one tracking-source /
    coordinate representation.

    This helper introduces no new scientific
    choices. It only consolidates repeated code.
    """

    # ------------------------------------------
    # Observed grouped-CV results
    # ------------------------------------------

    fold_results_by_feature = {}

    for (
        feature_name,
        dataset,
    ) in classifier_sets.items():

        fold_results = (
            evaluate_grouped_classifier(
                model=
                    make_phase8b_logistic_pipeline(),
                X=dataset["X"],
                y=dataset["y"],
                groups=dataset["groups"],
                cv=cv,
            )
        )

        fold_results_by_feature[
            feature_name
        ] = fold_results


    summary = pd.DataFrame(
        [
            {
                "feature_space":
                    feature_name,

                "mean_roc_auc":
                    float(
                        fold_results[
                            "roc_auc"
                        ].mean()
                    ),

                "sd_roc_auc":
                    float(
                        fold_results[
                            "roc_auc"
                        ].std(
                            ddof=1
                        )
                    ),

                "mean_balanced_accuracy":
                    float(
                        fold_results[
                            "balanced_accuracy"
                        ].mean()
                    ),

                "sd_balanced_accuracy":
                    float(
                        fold_results[
                            "balanced_accuracy"
                        ].std(
                            ddof=1
                        )
                    ),
            }
            for (
                feature_name,
                fold_results,
            ) in (
                fold_results_by_feature.items()
            )
        ]
    )


    observed_auc = {
        feature_name:
            get_mean_auc(
                summary,
                feature_name,
            )
        for feature_name in (
            expected_feature_dimensions.keys()
        )
    }


    observed_contrasts = pd.DataFrame(
        [
            {
                "contrast":
                    "odd_combined_minus_net",

                "observed_difference":
                    (
                        observed_auc[
                            "odd_combined"
                        ]
                        - observed_auc[
                            "net_only"
                        ]
                    ),
            },
            {
                "contrast":
                    "full_minus_net",

                "observed_difference":
                    (
                        observed_auc[
                            "full_6d"
                        ]
                        - observed_auc[
                            "net_only"
                        ]
                    ),
            },
            {
                "contrast":
                    "full_minus_odd_combined",

                "observed_difference":
                    (
                        observed_auc[
                            "full_6d"
                        ]
                        - observed_auc[
                            "odd_combined"
                        ]
                    ),
            },
        ]
    )


    # ------------------------------------------
    # Paired orientation-randomization null
    # ------------------------------------------

    rng = np.random.default_rng(
        null_seed
    )

    null_records = []


    for null_id in range(
        n_null
    ):

        swap_mask = (
            rng.random(
                len(
                    sequence_groups
                )
            )
            < 0.5
        )

        null_auc = {}

        for feature_name in (
            expected_feature_dimensions.keys()
        ):

            (
                pseudo_forward,
                pseudo_reverse,
            ) = randomize_pair_orientation(
                forward_spaces[
                    feature_name
                ],
                reverse_spaces[
                    feature_name
                ],
                swap_mask,
            )

            (
                X_null,
                y_null,
                groups_null,
            ) = build_classification_dataset(
                pseudo_forward,
                pseudo_reverse,
                sequence_groups,
            )

            fold_results = (
                evaluate_grouped_classifier(
                    model=
                        make_phase8b_logistic_pipeline(),
                    X=X_null,
                    y=y_null,
                    groups=groups_null,
                    cv=cv,
                )
            )

            null_auc[
                feature_name
            ] = float(
                fold_results[
                    "roc_auc"
                ].mean()
            )

        null_records.append(
            {
                "null_id":
                    null_id,

                "net_only_auc":
                    null_auc[
                        "net_only"
                    ],

                "internal_odd_only_auc":
                    null_auc[
                        "internal_odd_only"
                    ],

                "even_only_auc":
                    null_auc[
                        "even_only"
                    ],

                "odd_combined_auc":
                    null_auc[
                        "odd_combined"
                    ],

                "full_6d_auc":
                    null_auc[
                        "full_6d"
                    ],

                "odd_combined_minus_net":
                    (
                        null_auc[
                            "odd_combined"
                        ]
                        - null_auc[
                            "net_only"
                        ]
                    ),

                "full_minus_net":
                    (
                        null_auc[
                            "full_6d"
                        ]
                        - null_auc[
                            "net_only"
                        ]
                    ),

                "full_minus_odd_combined":
                    (
                        null_auc[
                            "full_6d"
                        ]
                        - null_auc[
                            "odd_combined"
                        ]
                    ),
            }
        )


    null_table = pd.DataFrame(
        null_records
    )


    # ------------------------------------------
    # Feature-space null summary
    # ------------------------------------------

    feature_null_records = []

    for feature_name in (
        expected_feature_dimensions.keys()
    ):

        observed = (
            observed_auc[
                feature_name
            ]
        )

        null_values = (
            null_table[
                null_column_by_feature[
                    feature_name
                ]
            ].to_numpy()
        )

        null_q95 = float(
            np.quantile(
                null_values,
                0.95,
            )
        )

        feature_null_records.append(
            {
                "feature_space":
                    feature_name,

                "observed_auc":
                    observed,

                "null_mean":
                    float(
                        np.mean(
                            null_values
                        )
                    ),

                "null_q95":
                    null_q95,

                "empirical_upper_p":
                    empirical_upper_pvalue(
                        observed,
                        null_values,
                    ),

                "above_null_q95":
                    bool(
                        observed
                        > null_q95
                    ),
            }
        )


    feature_null_summary = (
        pd.DataFrame(
            feature_null_records
        )
    )


    # ------------------------------------------
    # Contrast-null summary
    # ------------------------------------------

    observed_contrast_lookup = {
        row["contrast"]:
            float(
                row[
                    "observed_difference"
                ]
            )
        for _, row in (
            observed_contrasts.iterrows()
        )
    }


    contrast_records = []

    for (
        contrast_name,
        null_column,
    ) in (
        contrast_null_columns.items()
    ):

        observed = (
            observed_contrast_lookup[
                contrast_name
            ]
        )

        null_values = (
            null_table[
                null_column
            ].to_numpy()
        )

        null_q95 = float(
            np.quantile(
                null_values,
                0.95,
            )
        )

        contrast_records.append(
            {
                "contrast":
                    contrast_name,

                "observed_difference":
                    observed,

                "null_mean":
                    float(
                        np.mean(
                            null_values
                        )
                    ),

                "null_q95":
                    null_q95,

                "empirical_upper_p":
                    empirical_upper_pvalue(
                        observed,
                        null_values,
                    ),

                "above_null_q95":
                    bool(
                        observed
                        > null_q95
                    ),
            }
        )


    contrast_null_summary = (
        pd.DataFrame(
            contrast_records
        )
    )


    return {
        "fold_results":
            fold_results_by_feature,

        "summary":
            summary,

        "observed_contrasts":
            observed_contrasts,

        "null_table":
            null_table,

        "feature_null_summary":
            feature_null_summary,

        "contrast_null_summary":
            contrast_null_summary,
    }
```

The transformation is invertible.

```python
def validate_four_step_parity(
    forward_array,
):
    reverse_array = (
        reverse_k_step_sequences(
            forward_array
        )
    )

    forward_parity = (
        decompose_four_step_parity(
            forward_array
        )
    )

    reverse_parity = (
        decompose_four_step_parity(
            reverse_array
        )
    )

    (
        n_forward,
        q_forward,
        e1_forward,
        e2_forward,
    ) = forward_parity

    (
        n_reverse,
        q_reverse,
        e1_reverse,
        e2_reverse,
    ) = reverse_parity


    assert np.allclose(
        n_reverse,
        -n_forward,
    )

    assert np.allclose(
        q_reverse,
        -q_forward,
    )

    assert np.allclose(
        e1_reverse,
        e1_forward,
    )

    assert np.allclose(
        e2_reverse,
        e2_forward,
    )


    reconstructed = (
        reconstruct_four_step_from_parity(
            *forward_parity
        )
    )

    assert np.allclose(
        reconstructed,
        forward_array,
    )


    max_roundtrip_error = float(
        np.max(
            np.abs(
                reconstructed
                - forward_array
            )
        )
    )


    return {
        "net_odd":
            n_forward,

        "internal_odd":
            q_forward,

        "even_outer":
            e1_forward,

        "even_inner":
            e2_forward,

        "reverse_array":
            reverse_array,

        "max_roundtrip_error":
            max_roundtrip_error,
    }
```

### Four-step feature spaces

Before any four-step classifier performance is inspected, the exact
four-step feature matrices are fixed.

The five pre-specified feature spaces are:

1. `net_only`

$$
\mathbf{n}_4
$$

Dimension: 2.

2. `internal_odd_only`

$$
\mathbf{q}_4
$$

Dimension: 2.

3. `even_only`

$$
[
\mathbf{e}_{4,1},
\mathbf{e}_{4,2}
]
$$

Dimension: 4.

Both even coordinates are invariant under reversal, so the complete
`even_only` forward and reversed feature matrices must be identical.

4. `odd_combined`

$$
[
\mathbf{n}_4,
\mathbf{q}_4
]
$$

Dimension: 4.

5. `full_8d`

$$
[
\mathbf{d}_1,
\mathbf{d}_2,
\mathbf{d}_3,
\mathbf{d}_4
]
$$

Dimension: 8.

The original physical displacement coordinates are retained for the full
representation.

No alternative four-step feature representation will be selected after
classifier outcomes are inspected.

```python
FOUR_STEP_FEATURE_DIMENSIONS = {
    "net_only": 2,
    "internal_odd_only": 2,
    "even_only": 4,
    "odd_combined": 4,
    "full_8d": 8,
}


def make_four_step_feature_spaces(
    forward_array,
):
    reverse_array = (
        reverse_k_step_sequences(
            forward_array
        )
    )

    (
        n_forward,
        q_forward,
        e1_forward,
        e2_forward,
    ) = decompose_four_step_parity(
        forward_array
    )

    (
        n_reverse,
        q_reverse,
        e1_reverse,
        e2_reverse,
    ) = decompose_four_step_parity(
        reverse_array
    )


    forward_spaces = {
        "net_only":
            n_forward,

        "internal_odd_only":
            q_forward,

        "even_only":
            np.column_stack(
                [
                    e1_forward,
                    e2_forward,
                ]
            ),

        "odd_combined":
            np.column_stack(
                [
                    n_forward,
                    q_forward,
                ]
            ),

        "full_8d":
            forward_array,
    }


    reverse_spaces = {
        "net_only":
            n_reverse,

        "internal_odd_only":
            q_reverse,

        "even_only":
            np.column_stack(
                [
                    e1_reverse,
                    e2_reverse,
                ]
            ),

        "odd_combined":
            np.column_stack(
                [
                    n_reverse,
                    q_reverse,
                ]
            ),

        "full_8d":
            reverse_array,
    }


    return (
        forward_spaces,
        reverse_spaces,
    )
```

## Primary four-step result — CTC co-moving logistic analysis

The pre-specified CTC co-moving four-step logistic analysis did not satisfy the
criterion for higher-order temporal information beyond net displacement.

Observed mean ROC AUC values were:

- `internal_odd_only`: 0.496453

- `odd_combined`: 0.478387

None of the five feature spaces exceeded its paired

The internal reversal-odd coordinate was essentially at chance and did not

This is a null-calibrated non-detection and is not evidence that the underlying

```python
def run_four_step_logistic_analysis(
    forward_spaces,
    reverse_spaces,
    sequence_groups,
    classifier_sets,
    cv,
    null_seed=2029,
    n_null=200,
):
    fold_results_by_feature = {}

    for feature_name, dataset in classifier_sets.items():
        fold_results_by_feature[feature_name] = (
            evaluate_grouped_classifier(
                model=make_phase8b_logistic_pipeline(),
                X=dataset["X"],
                y=dataset["y"],
                groups=dataset["groups"],
                cv=cv,
            )
        )

    summary = pd.DataFrame(
        [
            {
                "feature_space": feature_name,
                "mean_roc_auc": float(results["roc_auc"].mean()),
                "sd_roc_auc": float(results["roc_auc"].std(ddof=1)),
                "mean_balanced_accuracy": float(
                    results["balanced_accuracy"].mean()
                ),
                "sd_balanced_accuracy": float(
                    results["balanced_accuracy"].std(ddof=1)
                ),
            }
            for feature_name, results
            in fold_results_by_feature.items()
        ]
    )

    observed_auc = {
        feature_name: get_mean_auc(
            summary,
            feature_name,
        )
        for feature_name
        in FOUR_STEP_FEATURE_DIMENSIONS.keys()
    }

    observed_contrasts = pd.DataFrame(
        [
            {
                "contrast": "odd_combined_minus_net",
                "observed_difference":
                    observed_auc["odd_combined"]
                    - observed_auc["net_only"],
            },
            {
                "contrast": "full_minus_net",
                "observed_difference":
                    observed_auc["full_8d"]
                    - observed_auc["net_only"],
            },
            {
                "contrast": "full_minus_odd_combined",
                "observed_difference":
                    observed_auc["full_8d"]
                    - observed_auc["odd_combined"],
            },
        ]
    )

    rng = np.random.default_rng(
        null_seed
    )

    null_records = []

    for null_id in range(n_null):

        swap_mask = (
            rng.random(
                len(sequence_groups)
            )
            < 0.5
        )

        null_auc = {}

        for feature_name in (
            FOUR_STEP_FEATURE_DIMENSIONS.keys()
        ):

            (
                pseudo_forward,
                pseudo_reverse,
            ) = randomize_pair_orientation(
                forward_spaces[
                    feature_name
                ],
                reverse_spaces[
                    feature_name
                ],
                swap_mask,
            )

            (
                X_null,
                y_null,
                groups_null,
            ) = build_classification_dataset(
                pseudo_forward,
                pseudo_reverse,
                sequence_groups,
            )

            fold_results = (
                evaluate_grouped_classifier(
                    model=
                        make_phase8b_logistic_pipeline(),
                    X=X_null,
                    y=y_null,
                    groups=groups_null,
                    cv=cv,
                )
            )

            null_auc[feature_name] = float(
                fold_results[
                    "roc_auc"
                ].mean()
            )

        null_records.append(
            {
                "null_id": null_id,
                "net_only_auc":
                    null_auc["net_only"],
                "internal_odd_only_auc":
                    null_auc["internal_odd_only"],
                "even_only_auc":
                    null_auc["even_only"],
                "odd_combined_auc":
                    null_auc["odd_combined"],
                "full_8d_auc":
                    null_auc["full_8d"],
                "odd_combined_minus_net":
                    null_auc["odd_combined"]
                    - null_auc["net_only"],
                "full_minus_net":
                    null_auc["full_8d"]
                    - null_auc["net_only"],
                "full_minus_odd_combined":
                    null_auc["full_8d"]
                    - null_auc["odd_combined"],
            }
        )

    null_table = pd.DataFrame(
        null_records
    )

    feature_null_records = []

    for feature_name in (
        FOUR_STEP_FEATURE_DIMENSIONS.keys()
    ):

        observed = observed_auc[
            feature_name
        ]

        null_values = null_table[
            FOUR_STEP_NULL_COLUMN_BY_FEATURE[
                feature_name
            ]
        ].to_numpy()

        q95 = float(
            np.quantile(
                null_values,
                0.95,
            )
        )

        feature_null_records.append(
            {
                "feature_space":
                    feature_name,
                "observed_auc":
                    observed,
                "null_mean":
                    float(
                        np.mean(
                            null_values
                        )
                    ),
                "null_q95":
                    q95,
                "empirical_upper_p":
                    empirical_upper_pvalue(
                        observed,
                        null_values,
                    ),
                "above_null_q95":
                    bool(
                        observed > q95
                    ),
            }
        )

    feature_null_summary = pd.DataFrame(
        feature_null_records
    )

    contrast_lookup = {
        row["contrast"]:
            float(
                row[
                    "observed_difference"
                ]
            )
        for _, row in (
            observed_contrasts.iterrows()
        )
    }

    contrast_records = []

    for contrast_name, null_column in (
        FOUR_STEP_CONTRAST_NULL_COLUMNS.items()
    ):

        observed = contrast_lookup[
            contrast_name
        ]

        null_values = null_table[
            null_column
        ].to_numpy()

        q95 = float(
            np.quantile(
                null_values,
                0.95,
            )
        )

        contrast_records.append(
            {
                "contrast":
                    contrast_name,
                "observed_difference":
                    observed,
                "null_mean":
                    float(
                        np.mean(
                            null_values
                        )
                    ),
                "null_q95":
                    q95,
                "empirical_upper_p":
                    empirical_upper_pvalue(
                        observed,
                        null_values,
                    ),
                "above_null_q95":
                    bool(
                        observed > q95
                    ),
            }
        )

    return {
        "summary":
            summary,
        "observed_contrasts":
            observed_contrasts,
        "null_table":
            null_table,
        "feature_null_summary":
            pd.DataFrame(
                feature_null_records
            ),
        "contrast_null_summary":
            pd.DataFrame(
                contrast_records
            ),
    }
```
